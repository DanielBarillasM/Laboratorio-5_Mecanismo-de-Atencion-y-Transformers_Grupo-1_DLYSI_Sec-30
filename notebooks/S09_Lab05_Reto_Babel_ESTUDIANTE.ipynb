### 1. ¿Qué limitación de los modelos recurrentes buscaba resolver el Transformer y por qué la traducción era una tarea importante para demostrarlo?

Los modelos recurrentes generan el estado de la posición $t$ a partir del estado $t-1$. Esta dependencia impone una cadena de cómputo: aunque exista una GPU capaz de procesar muchas operaciones simultáneamente, el estado posterior no puede calcularse hasta disponer del anterior. Vaswani et al. (2017) señalan que esta naturaleza secuencial limita la paralelización dentro de cada ejemplo y se vuelve especialmente costosa con secuencias largas, porque también restringe el tamaño efectivo de los lotes. Además, la ruta que conecta palabras distantes atraviesa múltiples pasos recurrentes, lo que dificulta representar dependencias lejanas.

El Transformer reemplaza esa recurrencia por atención: cada posición puede relacionarse directamente con otras posiciones relevantes. Durante el entrenamiento, todas las posiciones conocidas de una oración pueden procesarse en paralelo, aunque la generación final continúe siendo autorregresiva. La traducción fue una demostración exigente porque no basta clasificar una oración: el modelo debe construir una representación del mensaje de entrada, alinear elementos entre idiomas y producir otra secuencia respetando dependencias gramaticales. En Reto Babel esto se observa de forma controlada: el traductor debe reconocer que una frase con orden sujeto–objeto–verbo debe reorganizarse al patrón del español, además de mover la negación y preservar concordancia léxica.

**Fuente principal:** [Vaswani et al. (2017), secciones 1 y 3](https://arxiv.org/pdf/1706.03762)

### 2. ¿Qué trabajo realizan el encoder, la self-attention enmascarada del decoder y la atención entre decoder y encoder?

El **encoder** recibe todos los tokens de la frase secreta y produce una secuencia de representaciones contextualizadas. Su self-attention permite que cada posición combine información de cualquier otra posición de la entrada; por ello, el vector de una palabra puede incorporar su sujeto, objeto, adjetivo o marcador de negación aunque estén separados.

El **decoder** genera español de izquierda a derecha. Su self-attention debe estar enmascarada: al predecir la palabra $y_t$, solo puede consultar $y_1,\ldots,y_{t-1}$. La máscara causal coloca conexiones prohibidas sobre las posiciones futuras:

$$
M_{ij}=\begin{cases}0,&j\le i\\-\infty,&j>i.\end{cases}
$$

Sin esa restricción, durante el entrenamiento el decoder podría copiar información de la respuesta futura y la pérdida sería artificialmente optimista. Finalmente, la **atención encoder–decoder** utiliza consultas procedentes del decoder y claves/valores producidos por el encoder. Así, al generar cada palabra española, el decoder puede localizar qué partes de la frase secreta son pertinentes. PyTorch encapsula estas operaciones en `nn.Transformer`; nuestro trabajo consiste en suministrar embeddings, máscaras y tensores con dimensiones correctas, no en reprogramar la atención multi-cabeza.

**Fuentes:** [Vaswani et al. (2017), secciones 3.1 y 3.2](https://arxiv.org/pdf/1706.03762) · [Documentación oficial de `torch.nn.Transformer`](https://docs.pytorch.org/docs/stable/generated/torch.nn.Transformer.html)

### 3. ¿Por qué necesita información posicional y qué se espera si se elimina?

La atención por sí sola relaciona contenidos, pero no incorpora automáticamente el índice de cada token. Si se permutan los embeddings y no se añade ninguna señal de posición, el bloque no dispone de una noción explícita de “primero”, “después” o “último”. Esto es crítico en traducción, donde el orden altera la función sintáctica y el significado. El artículo suma a cada embedding una codificación sinusoidal de igual dimensión:

$$
PE_{(pos,2i)}=\sin\!\left(\frac{pos}{10000^{2i/d_{modelo}}}\right),\qquad
PE_{(pos,2i+1)}=\cos\!\left(\frac{pos}{10000^{2i/d_{modelo}}}\right).
$$

Estas señales permiten distinguir posiciones y ofrecen patrones relativos que el modelo puede combinar linealmente. Mi predicción previa es que el modelo **sin posición** perderá exactitud por token y, sobre todo, frases completamente correctas: podrá aprender asociaciones del diccionario y regularidades frecuentes, pero tendrá más dificultad para decidir qué palabra ocupa cada lugar después de transformar sujeto–objeto–verbo, adjetivos y negación. Como las frases son cortas y la gramática es muy regular, la caída podría no ser absoluta; precisamente por eso la ablación debe medirse en lugar de asumirse.

**Fuente:** [Vaswani et al. (2017), sección 3.5](https://arxiv.org/pdf/1706.03762)

### 4. ¿Qué ventaja y qué costo presenta la atención frente a la recurrencia, y cómo se relacionan con Reto Babel?

La ventaja principal es computacional y representacional. En una capa de self-attention, las posiciones pueden procesarse simultáneamente y la ruta máxima entre dos posiciones es constante. Esto favorece la paralelización y facilita relacionar elementos distantes. El costo es que la matriz de atención compara pares de posiciones y su complejidad respecto del largo es cuadrática:

$$
\mathcal{O}(n^2d),
$$

mientras una capa recurrente presenta $\mathcal{O}(nd^2)$, pero exige $\mathcal{O}(n)$ operaciones secuenciales. Para secuencias muy largas, almacenar y calcular $n^2$ relaciones puede dominar memoria y tiempo. En Reto Babel las frases son cortas, así que el costo cuadrático es pequeño; a la vez, la ventaja de paralelización será menos espectacular que en corpus de oraciones largas. El beneficio más importante aquí es didáctico y estructural: la atención permite inspeccionar la traducción como relaciones globales entre tokens, mientras las máscaras garantizan una generación causal.

**Fuente:** [Vaswani et al. (2017), sección 4 y tabla 1](https://arxiv.org/pdf/1706.03762)

```python
def leer_pares(ruta: Path):
    with ruta.open(encoding="utf-8", newline="") as archivo:
        return list(csv.DictReader(archivo))


class Vocabulario:
    ESPECIALES = ["<PAD>", "<SOS>", "<EOS>", "<UNK>"]

    def __init__(self, tokens):
        unicos = self.ESPECIALES + sorted(set(tokens) - set(self.ESPECIALES))
        self.token_a_id = {token: i for i, token in enumerate(unicos)}
        self.id_a_token = {i: token for token, i in self.token_a_id.items()}

    def __len__(self):
        return len(self.token_a_id)

    def codificar(self, texto, agregar_inicio_fin=True):
        ids = [self.token_a_id.get(token, self.token_a_id["<UNK>"]) for token in texto.split()]
        if agregar_inicio_fin:
            ids = [self.token_a_id["<SOS>"]] + ids + [self.token_a_id["<EOS>"]]
        return ids

    def decodificar(self, ids):
        tokens = []
        for indice in ids:
            token = self.id_a_token[int(indice)]
            if token == "<EOS>":
                break
            if token not in {"<PAD>", "<SOS>"}:
                tokens.append(token)
        return " ".join(tokens)


class ParesTraduccion(Dataset):
    def __init__(self, pares, vocab_src, vocab_tgt):
        self.pares = pares
        self.vocab_src = vocab_src
        self.vocab_tgt = vocab_tgt

    def __len__(self):
        return len(self.pares)

    def __getitem__(self, indice):
        fila = self.pares[indice]
        src = torch.tensor(self.vocab_src.codificar(fila["idioma_secreto"]), dtype=torch.long)
        tgt = torch.tensor(self.vocab_tgt.codificar(fila["espanol"]), dtype=torch.long)
        return src, tgt
```

<div class="lab-section"><h2>2 · Lotes, padding y máscaras</h2></div>
<div class="lab-call">El padding crea tensores rectangulares; las máscaras impiden atender a relleno y preservan la causalidad del decoder.</div>

Para un lote de secuencias con longitudes distintas se produce $X\in\mathbb{N}^{B\times L_{max}}$. La máscara de padding tiene la misma forma y vale `True` en `<PAD>`. La máscara causal es cuadrada, de tamaño $L_{tgt}\times L_{tgt}$, y oculta estrictamente la región superior a la diagonal.


```python
def crear_collate(pad_src, pad_tgt):
    """Crea una función de colación que conserva lotes como (batch, secuencia)."""
    def collate(lote):
        srcs, tgts = zip(*lote)
        src_lote = pad_sequence(
            srcs, batch_first=True, padding_value=pad_src
        )
        tgt_lote = pad_sequence(
            tgts, batch_first=True, padding_value=pad_tgt
        )
        return src_lote, tgt_lote
    return collate


def crear_mascara_causal(tamano, device):
    """Oculta las posiciones futuras del decoder mediante una máscara booleana."""
    return torch.triu(
        torch.ones((tamano, tamano), dtype=torch.bool, device=device),
        diagonal=1,
    )


def crear_mascara_padding(tokens, pad_idx):
    """Marca con True cada posición de padding que debe ignorar la atención."""
    return tokens.eq(pad_idx)
```

```python
PAD_SRC = vocab_src.token_a_id["<PAD>"]
PAD_TGT = vocab_tgt.token_a_id["<PAD>"]
collate_fn = crear_collate(PAD_SRC, PAD_TGT)

ds_entrenamiento = ParesTraduccion(pares_entrenamiento, vocab_src, vocab_tgt)
ds_validacion = ParesTraduccion(pares_validacion, vocab_src, vocab_tgt)
dl_entrenamiento = DataLoader(ds_entrenamiento, batch_size=64, shuffle=True, collate_fn=collate_fn)
dl_validacion = DataLoader(ds_validacion, batch_size=128, shuffle=False, collate_fn=collate_fn)

src_prueba, tgt_prueba = next(iter(dl_entrenamiento))
mascara = crear_mascara_causal(tgt_prueba.size(1) - 1, torch.device("cpu"))

assert src_prueba.ndim == 2 and tgt_prueba.ndim == 2
assert mascara.shape == (tgt_prueba.size(1) - 1, tgt_prueba.size(1) - 1)
assert mascara.dtype == torch.bool
assert not mascara.diag().any() and mascara[0, -1]
assert crear_mascara_padding(src_prueba, PAD_SRC).shape == src_prueba.shape
print("OK - lotes y máscaras correctos", src_prueba.shape, tgt_prueba.shape)

# Ejemplos de funciones utilizadas en este bloque:
# DataLoader(...), next(iter(...)), Tensor.size(...), torch.device(...)
```

<div class="lab-call gold"><b>Comprobación conceptual.</b> La diagonal permanece visible porque una posición puede usar su representación actual. Solo las columnas futuras quedan enmascaradas. Las máscaras de padding y causal resuelven problemas distintos y ambas son necesarias.</div>

<div class="lab-section"><h2>3 · Traductor con nn.Transformer</h2></div>
<div class="lab-call">Se conectan embeddings, codificación posicional, encoder–decoder de PyTorch y una proyección al vocabulario español.</div>

Los embeddings se escalan por $\sqrt{d_{modelo}}$ antes de sumar posición. Con `batch_first=True`, las entradas siguen la convención $(B,L,d)$. La capa final transforma cada estado del decoder en logits sobre $|V_{tgt}|$ posibles palabras.


```python
class CodificacionPosicional(nn.Module):
    """Codificación sinusoidal del artículo original, ya implementada."""

    def __init__(self, d_modelo, max_largo=64, dropout=0.1):
        super().__init__()
        self.dropout = nn.Dropout(dropout)
        posicion = torch.arange(max_largo, dtype=torch.float32).unsqueeze(1)
        divisor = torch.exp(
            torch.arange(0, d_modelo, 2, dtype=torch.float32)
            * (-math.log(10000.0) / d_modelo)
        )
        pe = torch.zeros(1, max_largo, d_modelo)
        pe[0, :, 0::2] = torch.sin(posicion * divisor)
        pe[0, :, 1::2] = torch.cos(posicion * divisor)
        self.register_buffer("pe", pe)

    def forward(self, x):
        return self.dropout(x + self.pe[:, :x.size(1)])
```

```python
class TraductorTransformer(nn.Module):
    def __init__(self, n_src, n_tgt, pad_src, pad_tgt, d_modelo=48, cabezas=4,
                 capas=2, d_ff=96, dropout=0.1, usar_posicion=True):
        super().__init__()
        self.d_modelo = d_modelo
        self.pad_src = pad_src
        self.pad_tgt = pad_tgt
        self.usar_posicion = usar_posicion

        self.embedding_src = nn.Embedding(n_src, d_modelo, padding_idx=pad_src)
        self.embedding_tgt = nn.Embedding(n_tgt, d_modelo, padding_idx=pad_tgt)
        self.posicion = CodificacionPosicional(d_modelo, max_largo=64, dropout=dropout)
        self.transformer = nn.Transformer(
            d_model=d_modelo,
            nhead=cabezas,
            num_encoder_layers=capas,
            num_decoder_layers=capas,
            dim_feedforward=d_ff,
            dropout=dropout,
            batch_first=True,
        )
        self.salida = nn.Linear(d_modelo, n_tgt)

    def forward(self, src, tgt_entrada):
        escala = math.sqrt(self.d_modelo)
        src_emb = self.embedding_src(src) * escala
        tgt_emb = self.embedding_tgt(tgt_entrada) * escala

        if self.usar_posicion:
            src_emb = self.posicion(src_emb)
            tgt_emb = self.posicion(tgt_emb)

        mascara_tgt = crear_mascara_causal(tgt_entrada.size(1), tgt_entrada.device)
        padding_src = crear_mascara_padding(src, self.pad_src)
        padding_tgt = crear_mascara_padding(tgt_entrada, self.pad_tgt)

        estados = self.transformer(
            src=src_emb,
            tgt=tgt_emb,
            tgt_mask=mascara_tgt,
            src_key_padding_mask=padding_src,
            tgt_key_padding_mask=padding_tgt,
            memory_key_padding_mask=padding_src,
        )
        return self.salida(estados)
```

```python
modelo_prueba = TraductorTransformer(
    len(vocab_src), len(vocab_tgt), PAD_SRC, PAD_TGT, dropout=0.0
)
logits_prueba = modelo_prueba(src_prueba[:3], tgt_prueba[:3, :-1])

assert logits_prueba.shape == (3, tgt_prueba.size(1) - 1, len(vocab_tgt))
assert sum(p.numel() for p in modelo_prueba.parameters()) < 250_000
print("OK - modelo conectado; parámetros:", sum(p.numel() for p in modelo_prueba.parameters()))

# Ejemplos de funciones utilizadas en este bloque:
# TraductorTransformer(...), Tensor.numel(), model.parameters()
```

<div class="lab-section"><h2>4 · Entrenamiento con teacher forcing</h2></div>
<div class="lab-call">El decoder recibe el objetivo desplazado; la entropía cruzada ignora `<PAD>` y el gradiente se recorta para estabilizar la optimización.</div>

$$
\mathcal{L}=-\frac{1}{N}\sum_{t:y_t\ne PAD}\log p_\theta(y_t\mid y_{<t},x).
$$

La pérdida de cada lote se pondera por su cantidad real de tokens para que los lotes con más padding no distorsionen el promedio de la época.


```python
def ejecutar_epoca(modelo, loader, criterio, optimizador=None):
    """Ejecuta una época y devuelve pérdida media por token no padding."""
    entrenando = optimizador is not None
    modelo.train(entrenando)
    total_perdida = 0.0
    total_tokens = 0

    for src, tgt in loader:
        src, tgt = src.to(DISPOSITIVO), tgt.to(DISPOSITIVO)
        tgt_entrada = tgt[:, :-1]
        tgt_esperado = tgt[:, 1:]

        if entrenando:
            optimizador.zero_grad(set_to_none=True)

        with torch.set_grad_enabled(entrenando):
            logits = modelo(src, tgt_entrada)
            perdida = criterio(
                logits.reshape(-1, logits.size(-1)),
                tgt_esperado.reshape(-1),
            )
            if entrenando:
                perdida.backward()
                torch.nn.utils.clip_grad_norm_(modelo.parameters(), max_norm=1.0)
                optimizador.step()

        tokens_validos = tgt_esperado.ne(PAD_TGT).sum().item()
        total_perdida += perdida.item() * tokens_validos
        total_tokens += tokens_validos

    if total_tokens == 0:
        raise RuntimeError("El lote no contiene tokens objetivo distintos de <PAD>.")
    return total_perdida / total_tokens
```

```python
def entrenar_modelo(usar_posicion=True, epocas=22):
    torch.manual_seed(SEMILLA)
    modelo = TraductorTransformer(
        len(vocab_src), len(vocab_tgt), PAD_SRC, PAD_TGT,
        d_modelo=48, cabezas=4, capas=2, d_ff=96,
        dropout=0.1, usar_posicion=usar_posicion,
    ).to(DISPOSITIVO)
    criterio = nn.CrossEntropyLoss(ignore_index=PAD_TGT)
    optimizador = torch.optim.Adam(modelo.parameters(), lr=2e-3)
    historia = {"train": [], "val": []}
    mejor_estado = None
    mejor_val = float("inf")

    for epoca in range(1, epocas + 1):
        perdida_train = ejecutar_epoca(modelo, dl_entrenamiento, criterio, optimizador)
        perdida_val = ejecutar_epoca(modelo, dl_validacion, criterio)
        historia["train"].append(perdida_train)
        historia["val"].append(perdida_val)
        if perdida_val < mejor_val:
            mejor_val = perdida_val
            mejor_estado = {k: v.detach().cpu().clone() for k, v in modelo.state_dict().items()}
        if epoca == 1 or epoca % 5 == 0 or epoca == epocas:
            print(f"época {epoca:02d} | train {perdida_train:.4f} | val {perdida_val:.4f}")

    modelo.load_state_dict(mejor_estado)
    return modelo, historia


modelo, historia = entrenar_modelo(usar_posicion=True)
assert len(historia["train"]) == 22
assert historia["train"][-1] < historia["train"][0]
print("OK - entrenamiento completado")

plt.plot(historia["train"], label="entrenamiento")
plt.plot(historia["val"], label="validación")
plt.xlabel("Época")
plt.ylabel("Pérdida")
plt.legend()
plt.show()

# Ejemplos de funciones utilizadas en este bloque:
# nn.CrossEntropyLoss(...), torch.optim.Adam(...), model.state_dict()
# Tensor.detach().cpu().clone(), plt.plot(...)
```

<div class="lab-call gold"><b>Qué debe observarse.</b> Una curva de entrenamiento descendente confirma optimización, pero la selección se basa en la menor pérdida de validación. Una brecha creciente entre ambas curvas sugeriría sobreajuste.</div>

<div class="lab-section"><h2>5 · Traducción progresiva y evaluación</h2></div>
<div class="lab-call">La inferencia comienza con `<SOS>`, agrega el `argmax` del último paso y termina con `<EOS>` o el largo máximo.</div>

Se reportan dos métricas complementarias: **exactitud por token**, que concede crédito parcial, y **frases exactas**, que exige que toda la secuencia coincida. La segunda es más estricta y corresponde al primer criterio del concurso.


```python
@torch.no_grad()
def traducir(modelo, frase_secreta, max_largo=16):
    """Genera español de forma greedy hasta <EOS> o el límite indicado."""
    modelo.eval()
    src_ids = vocab_src.codificar(frase_secreta)
    src = torch.tensor([src_ids], dtype=torch.long, device=DISPOSITIVO)

    sos = vocab_tgt.token_a_id["<SOS>"]
    eos = vocab_tgt.token_a_id["<EOS>"]
    generados = [sos]

    for _ in range(max_largo):
        tgt_actual = torch.tensor([generados], dtype=torch.long, device=DISPOSITIVO)
        logits = modelo(src, tgt_actual)
        siguiente = logits[0, -1].argmax().item()
        generados.append(siguiente)
        if siguiente == eos:
            break

    return vocab_tgt.decodificar(generados)
```

```python
def evaluar_traducciones(modelo, pares, limite=None):
    seleccion = pares if limite is None else pares[:limite]
    tokens_correctos = 0
    tokens_totales = 0
    frases_exactas = 0
    for fila in seleccion:
        pred = traducir(modelo, fila["idioma_secreto"])
        real = fila["espanol"]
        frases_exactas += int(pred == real)
        pred_tokens, real_tokens = pred.split(), real.split()
        largo = max(len(pred_tokens), len(real_tokens))
        tokens_correctos += sum(
            i < len(pred_tokens) and i < len(real_tokens) and pred_tokens[i] == real_tokens[i]
            for i in range(largo)
        )
        tokens_totales += largo
    return {
        "exactitud_tokens": tokens_correctos / tokens_totales,
        "frases_exactas": frases_exactas / len(seleccion),
    }


metricas = evaluar_traducciones(modelo, pares_validacion)
print(metricas)
for fila in pares_validacion[:5]:
    print("secreto :", fila["idioma_secreto"])
    print("esperado:", fila["espanol"])
    print("modelo  :", traducir(modelo, fila["idioma_secreto"]), "\n")

assert 0.0 <= metricas["exactitud_tokens"] <= 1.0
assert metricas["exactitud_tokens"] >= 0.70, "El traductor todavía necesita entrenamiento o correcciones."
print("OK - traductor funcional")

# Ejemplos de funciones utilizadas en este bloque:
# traducir(...), str.split(), max(...), sum(...)
```
